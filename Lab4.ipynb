# Visualization Exercises

 Import Numpy, Panda, Matplotlib and any other necessary library.

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

Assign the following variables:

```python
x = np.arange(0,50)
y = x*2
z = x**2
```

Draw y and z versus x in the same plot. The plot must be compelete (contain lables, legend, etc.)

```python
plt.plot(x,y,'r',x,z,'b')
plt.gca().legend(['x*2','x**2'])
plt.xlabel('x')
plt.show()
```

Create a bar chart from the Age column of the dataset. 

```python
df.plot.bar(y='Age' ,x="PassengerId")
plt.show()
```

Create a pie chart from the Age column of the dataset.  

```python
counts = df['Age'].value_counts();
counts.plot(kind="pie", autopct = '%.1f%%')
plt.show()
```

Create a swarmplot of the Age and Sex of the dataset.

```python
titanic_data = sns.load_dataset('Titanic',data_home='/Users/patriciasiuhaliu/Documents/Semester 2/python/week 5/Lab 4/');
print(titanic_data)
sns.swarmplot(y = 'Age', x = 'Sex', data = titanic_data)
plt.show()
```

Plot heatmap of the dataset with corresponding annotations.


```python
corr_matrix = titanic_data.corr()
print(corr_matrix)
sns.heatmap(corr_matrix, annot = True)
plt.show()
```

Draw a scatter plot of the Age vs. Fare from the dataset. 

```python
plt.scatter(y = 'Fare', x = 'Age', data = titanic_data)
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```
