## Exercise 3

Find "Pearson correlation" and "Spearman correlation" between "Age" and "Parch" columns. Discuss your observations and comparision.

```python
# CODE HERE
titanic['Age'].fillna(titanic['Age'].mean(),inplace=True)
age = titanic['Age']
parch = titanic['Parch']
spearmanr_coefficient, p_value = spearmanr(age, parch)
print('Spearman Rank Correlation Coefficient %0.3f' % (spearmanr_coefficient))
pearsonr_coefficient, p_value = pearsonr(age, parch)
print('PeasonR Correlation Coefficient %0.3f'% (pearsonr_coefficient))
```

```python
#Spearman -Close to 0, is almost monotonic nrgative, not linear, 
#pearson - close to 0, negative low strength nature of relationship,
#almost not linear, scattered scenario

#Spearman shows higher correlation between age and parch copared to pearson
#Both show negative correlation
```
